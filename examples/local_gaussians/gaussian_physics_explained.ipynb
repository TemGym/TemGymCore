For a single ray locally representing a wavefront, we want to solve the equation

$$U'(\mathbf{r}')=\iint_{\mathbb{R}^2} U(\mathbf{r})\,K_S(\mathbf{r},\mathbf{r}')\,\mathrm{d}^2\mathbf{r}$$

where $\mathbf{r}$ and $\mathbf{r}'$ are the coordinates of the wavefront at the input and output plane, $U$ and $U'$ represent the complex amplitudes at the input and output planes, and $K_S(\mathbf{r},\mathbf{r}')$ is the kernel function that describes the propagation of the wavefront from the input to the output plane. One area element $\mathrm{d}^2\mathbf{r}$ in the integral represents a local complex patch of amplitude and phase, and the integral represents a sum of all distorted wavefronts from that little patch of area on an input, to all points on the output plane. Importantly by choosing to represent the propagation integral in this way, we are free to choose a propagation kernel $K_S$ that is not limited just to free-space propagation, or even paraxial propagation. The kernel can represent any propagation that can be described by a real or complex optical path length function $S(\mathbf{r},\mathbf{r}')$.

Unfortunately, solving this integral analytically is not possible for arbitrary wavefronts and arbitrary kernels, and solving it numerically in a brute force manner is computationally expensive. In order to make this integral more manageable, we have two levers we can choose to play with: Choose an amenable input wavefront $U(\mathbf{r})$ and choose an amenable kernel $K_S(\mathbf{r},\mathbf{r}')$.

We will initially concentrate on how to choose an amenable input wavefront $U(\mathbf{r})$.

The function we will choose to locally represent the wavefront is as follows:
Let the local expansion point (the center of the complex patch) be denoted by $\mathbf{r}_0=(x_0,y_0)$ and a generic point in the input plane by $\mathbf{r}=(x,y)$. Then the local wavefront can be written

$$
U(\mathbf{r}) = A\exp\big(ikS(\mathbf{r})\big),
$$

and the Taylor expansion of the action $S$ around $\mathbf{r}_0$ is

$$
S(\mathbf{r}) \approx S(\mathbf{r}_0) + \nabla S(\mathbf{r}_0)\cdot(\mathbf{r}-\mathbf{r}_0)
+\tfrac{1}{2}(\mathbf{r}-\mathbf{r}_0)^\top H_S(\mathbf{r}_0)\,(\mathbf{r}-\mathbf{r}_0),
$$

where
$$
\nabla S(\mathbf{r}_0)=\big(\partial_x S,\;\partial_y S\big)\big|_{\mathbf{r}_0},\qquad
H_S(\mathbf{r}_0)=\begin{pmatrix}\partial_x^2 S & \partial_x\partial_y S\\[2pt]\partial_y\partial_x S & \partial_y^2 S\end{pmatrix}\bigg|_{\mathbf{r}_0}.
$$

Substituting this expansion into $U$ gives the local quadratic-phase approximation

$$
U(\mathbf{r}) \approx A\,
\underbrace{e^{ikS(\mathbf{r}_0)}}_{\text{constant offset}}
\underbrace{e^{ik\nabla S(\mathbf{r}_0)\cdot(\mathbf{r}-\mathbf{r}_0)}}_{\text{linear (tilt) term}}
\underbrace{e^{\tfrac{ik}{2}(\mathbf{r}-\mathbf{r}_0)^\top H_S(\mathbf{r}_0)\,(\mathbf{r}-\mathbf{r}_0)}}_{\text{local curvature term}}
$$

The first term is a constant phase offset, the second term is a linear phase term that represents a tilt of the wavefront, and the third term is a quadratic phase term that represents the local curvature of the wavefront. The local curvature term can represent a converging or diverging wavefront, or a saddle point.

We can compare the local quadratic term in the above taylor expansion $H_S(\mathbf{r}_0)$ to gaussian beam solution of the paraxial wave equation, which has a quadratic phase term of the form with the so called complex curvature matrix $Q_{\text{inv}}$:

$$
U(\mathbf{r}) = e^{\tfrac{ik}{2}(\mathbf{r}-\mathbf{r}_0)^\top Q_{\text{inv}}(\mathbf{r}_0)\,(\mathbf{r}-\mathbf{r}_0)}
$$

where the complex curvature matrix is composed of a 2x2 matrix of complex curvature values $q_{xx}, q_{xy}, q_{yx}, q_{yy}$:

$$
Q_{\text{inv}}(\mathbf{r}_0) = \begin{pmatrix}q_{xx} & q_{xy}\\[2pt]q_{yx} & q_{yy}\end{pmatrix}\bigg|_{\mathbf{r}_0}
$$

The individual complex curvature value is defined as

$$q = z - jz_{R}$$

where $z$ is the axial position of the wavefront and $z_{R}$ is known as the Rayleigh range, a measure of how quickly the beam diverges.

Inverting the above equation, we can define two real value functions, $R(z)$ and $W(z)$, that represent the radius of curvature and the beam waist respectively:

$$
\frac{1}{q(z)} = \frac{1}{R(z)} + j\frac{\lambda}{\pi W^2(z)}
$$

where $\lambda$ is the wavelength of the electric wave. 


### Example: Thin Lens
Let $\mathbf{r}=(x,y)$ be the 2D coordinate and $\mathbf{r}_0=(x_0,y_0)$ the expansion point. For a thin (paraxial) lens with focal length f,

$$
S(\mathbf{r})=\frac{x^2+y^2}{2f}=\frac{\|\mathbf{r}\|^2}{2f}.
$$

Gradient and Hessian:
$$
\nabla S(\mathbf{r})=\frac{\mathbf{r}}{f},\qquad
H_S(\mathbf{r})=\frac{1}{f}I_{2},
$$

where \(I_2\) is the 2×2 identity matrix.

Second-order Taylor expansion about \mathbf{r}_0:
$$
S(\mathbf{r})\approx S(\mathbf{r}_0)+\nabla S(\mathbf{r}_0)\cdot(\mathbf{r}-\mathbf{r}_0)
+\tfrac{1}{2}(\mathbf{r}-\mathbf{r}_0)^\top H_S(\mathbf{r}_0)\,(\mathbf{r}-\mathbf{r}_0).
$$

Substituting the lens expressions:
$$
S(\mathbf{r})\approx \frac{\|\mathbf{r}_0\|^2}{2f}+\frac{\mathbf{r}_0}{f}\cdot(\mathbf{r}-\mathbf{r}_0)
+\frac{1}{2f}\|\mathbf{r}-\mathbf{r}_0\|^2.
$$

Remarks:
- Since $S$ is exactly quadratic, the expansion is exact.
- The local curvature matrix is $H_S(\mathbf{r}_0)=\tfrac{1}{f}I_2$, so in the quadratic-phase form the coefficient matrix is $\tfrac{ik}{2}H_S=\tfrac{ik}{2f}I_2$.
- The linear term $\frac{\mathbf{r}_0}{f}\cdot(\mathbf{r}-\mathbf{r}_0)$ represents the local tilt induced by the lens at $\mathbf{r}_0$.

$$
\begin{aligned}
S_{\text{const}}' &= S_{\text{const}} + \Delta S_0,\\
\mathbf{S}_{\text{lin}}' &= \mathbf{S}_{\text{lin}} + \Delta \mathbf{S}_1,\\
S_{\text{quad}}' &= S_{\text{quad}} + \Delta S_2,
\end{aligned}
$$

where
- $\Delta S_0 = S_{\text{kernel}}(\mathbf{r}_0)$,
- $\Delta \mathbf{S}_1 = \nabla S_{\text{kernel}}(\mathbf{r}_0)$,
- $\Delta S_2 = H_{S_{\text{kernel}}}(\mathbf{r}_0)$.

Therefore to update the gaussian beam parameters after propagation, we simply add the contributions from the kernel's Taylor expansion at the expansion point $\mathbf{r}_0$ to the existing parameters of the gaussian beam.

Let's start our example in code by creating a 1D gaussian ray object and visualising it's amplitude and phase, and how it changes as we manipulate the action variables.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.widgets import Slider
%matplotlib widget

class GaussianRay1D:
    """
    E(x) = C * exp(i k S(x)),   S(x) = S0 + g x + 0.5 H x^2 (all complex allowed)
    with k > 0 real, and C complex constant (gain/phase).
    """
    def __init__(self, C=1+0j, S0=0+0j, g=0+0j, H=1j, k=1.0):
        # Action parameters
        self.S0 = complex(S0)
        self.g  = complex(g)
        self.H  = complex(H)
        self.k  = float(k)

        # Global multiplicative constant (gain/phase)
        self.C  = complex(C)

    # ---------- Core action ----------
    def S(self, x):
        x = np.asarray(x)
        return self.S0 + self.g * x + 0.5 * self.H * x**2

    # ---------- Observables (respect Frozen C) ----------
    def amplitude(self, x):
        # |E| = |C| * exp(-k * Im S)
        return abs(self.C) * np.exp(-self.k * np.imag(self.S(x)))

    def intensity(self, x):
        a = self.amplitude(x)
        return a**2

    def phase(self, x):
        # arg(E) = arg(C) + k Re S
        return np.angle(self.C) + self.k * np.real(self.S(x))

    def tilt(self, x):
        # paraxial slope = (d/dx phase)/k = Re(g + H x)/k
        return np.real(self.g + self.H * x) / self.k

    def intensity_center(self):
        # x_c = -Im(g) / Im(H)  (requires Im(H) > 0 to be localized)
        HI = np.imag(self.H)
        gI = np.imag(self.g)
        if np.isclose(HI, 0.0):
            return np.nan
        return -gI / HI

    # ---------- Initial amplitude/phase setup (one-time) ----------
    def set_initial_peak_amplitude(self, A0):
        """
        Set |C| so that the initial peak field amplitude |E(x_c)| equals A0,
        given the current S parameters. Does NOT change later when S changes,
        because we keep C frozen.
        """
        if np.isclose(self.H.imag, 0.0):
            # Not confined; fall back to the value at x=0
            SIc = self.S0.imag
        else:
            SIc = self.S0.imag - (self.g.imag**2) / (2.0 * self.H.imag)
        # |E(x_c)| = |C| * exp(-k * S_I(x_c))  =>  |C| = A0 * exp(+k * S_I(x_c))
        mag = float(A0) * np.exp(self.k * SIc)
        # keep current global phase of C
        self.C = mag * np.exp(1j * np.angle(self.C))

    def set_initial_peak_intensity(self, I0):
        self.set_initial_peak_amplitude(np.sqrt(float(I0)))

    def set_initial_global_phase(self, theta):
        """Set arg(C) = theta (radians)."""
        self.C = abs(self.C) * np.exp(1j * float(theta))

    # ---------- Action updates (C stays frozen) ----------
    def update_action(self, *, S0=None, g=None, H=None, k=None):
        """
        Update any subset of (S0, g, H, k). Under the 'Frozen C' convention,
        this NEVER changes C—even if Im(S) changed and peak/power would shift.
        """
        if S0 is not None: self.S0 = complex(S0)
        if g  is not None: self.g  = complex(g)
        if H  is not None: self.H  = complex(H)
        if k  is not None: self.k  = float(k)
```

```python
ray = GaussianRay1D(S0=0+1j, g=0.3 + 0.0j, H=0.2 + 1.0j, C=1.0, k=1.0)

ray.set_initial_peak_amplitude(A0=1.0)   # sets |C| once
ray.set_initial_global_phase(theta=0.0)  # optional

x = np.linspace(-6, 6, 800)

# This is S_I evaluated at the envelope center (independent of x): stays constant.
HI0 = ray.H.imag
gI0 = ray.g.imag

# ---- Combined figure: Intensity and Phase side-by-side, sliders below ----
fig, (axI, axP) = plt.subplots(1, 2, figsize=(18, 8))

# Two-column slider layout; reserve ~24% of figure height
slider_h = 0.03
slider_gap = 0.006
rows = 7
start_y = 0.05
top_y = start_y + (rows - 1) * (slider_h + slider_gap) + slider_h
fig.subplots_adjust(bottom=top_y + 0.05)

# Intensity plot (left)
axI.set_title("Intensity profile I(x) and envelope center")
lineI, = axI.plot(x, ray.intensity(x), lw=2)
xc0 = ray.intensity_center()
if np.isfinite(xc0):
    center_dot, = axI.plot([xc0], [ray.intensity(xc0)], 'o', ms=8)
else:
    center_dot, = axI.plot([np.nan], [np.nan], 'o', ms=8)
axI.set_xlabel("x")
axI.set_ylabel("I(x)")
axI.grid(True)

# Phase plot (right) - show wrapped phase
axP.set_title("Phase profile φ(x) = k Re[S(x)]")
yP0 = ray.phase(x)
lineP_wrapped, = axP.plot(x, np.angle(np.exp(1j*yP0)))  # default solid
axP.set_xlabel("x")
axP.set_ylabel("φ(x)")
axP.grid(True)
axP.set_ylim(-np.pi, np.pi)

# ---- Slider axes in two columns ----
left = 0.10
col_w = 0.36
col_gap = 0.06
right = left + col_w + col_gap

def y_at(row):
    return start_y + row * (slider_h + slider_gap)

# Left column (4 rows)
ax_s_S0R = fig.add_axes([left, y_at(0), col_w, slider_h])
ax_s_S0I = fig.add_axes([left, y_at(1), col_w, slider_h])
ax_s_gR = fig.add_axes([left,  y_at(2), col_w, slider_h])
ax_s_gI = fig.add_axes([left,  y_at(3), col_w, slider_h])
ax_s_hR = fig.add_axes([left,  y_at(4), col_w, slider_h])
ax_s_hI = fig.add_axes([left,  y_at(5), col_w, slider_h])
ax_s_k   = fig.add_axes([left,  y_at(6), col_w, slider_h])

# Create sliders (Im(H) constrained >= 0 for a confined Gaussian)
s_S0R = Slider(ax_s_S0R,'Re(S0)', -5.0, 5.0, valinit=np.real(ray.S0))
s_S0I = Slider(ax_s_S0I,'Im(S0)', -2.0, 2.0, valinit=np.imag(ray.S0))
s_gR  = Slider(ax_s_gR, 'Re(g)',  -2.0, 2.0, valinit=np.real(ray.g))
s_gI  = Slider(ax_s_gI, 'Im(g)',  -2.0, 2.0, valinit=np.imag(ray.g))
s_hR  = Slider(ax_s_hR, 'Re(H)',  -2.0, 2.0, valinit=np.real(ray.H))
s_hI  = Slider(ax_s_hI, 'Im(H)',   0.0, 3.0, valinit=np.imag(ray.H))
s_k   = Slider(ax_s_k,  'k',       0.05, 5.0, valinit=ray.k)

# Text readout just above slider area
txt = fig.text(0.02, top_y + 0.01, "", transform=fig.transFigure, ha="left", va="bottom")

def update(_=None):
    # Read sliders
    g   = complex(s_gR.val,  s_gI.val)
    H   = complex(s_hR.val,  s_hI.val)
    k   = float(s_k.val)
    S0R = float(s_S0R.val)
    S0I = float(s_S0I.val)

    # Apply back to the ray (C stays frozen elsewhere)
    ray.g  = g
    ray.H  = H
    ray.k  = k
    ray.S0 = complex(S0R, S0I)   # <-- no invariant correction

    # Recompute & redraw
    yI = ray.amplitude(x)
    yP_wrapped = np.angle(np.exp(1j * ray.phase(x)))
    lineI.set_ydata(yI)
    lineP_wrapped.set_ydata(yP_wrapped)

    xc = ray.intensity_center()
    if np.isfinite(xc):
        center_dot.set_data([xc], [ray.amplitude(xc)])
    else:
        center_dot.set_data([np.nan], [np.nan])

    axI.relim(); axI.autoscale_view()
    axP.set_ylim(-np.pi, np.pi)
    fig.canvas.draw_idle()

for s in (s_gR, s_gI, s_hR, s_hI, s_k, s_S0R, s_S0I):
    s.on_changed(update)

# initial draw
update()
axP.legend(loc='upper right')
plt.show()
```
